# 🧠 Multi-Method Sentiment & Emotion Analysis Pipeline

**Methods covered:**
1. **Bayesian Naïve Bayes** — log-posterior ratio with a political/diplomatic lexicon  
2. **BERT Sentiment** — `cardiffnlp/twitter-roberta-base-sentiment-latest`  
3. **BERT Ekman Emotions** — `j-hartmann/emotion-english-distilroberta-base`  
4. **GPT Sentiment + Ekman** — OpenAI API (`gpt-4o-mini`) with structured JSON output  
5. **Sentence-level analysis** — Bayesian + BERT per sentence  
6. **Visualisations** — bar charts, radar chart, heatmap  

> **Text analysed:** Marco Rubio / Donald Trump / Iran diplomatic negotiations excerpt (March 2025)

## 2. Source Text & Sentence Splitting

````python
TEXT = (
    'US Secretary of State Marco Rubio says the administration must prepare for the '
    '"probability" that negotiations could fail, after Donald Trump said talks were '
    'happening with a "more reasonable" Iranian regime. Speaking to ABC\'s Good Morning '
    'America, Rubio would not reveal who the US is speaking to, saying, "it would '
    'probably get them in trouble with some other groups inside of Iran". '
    '"But at the end of the day we have to see if these people end up being the ones '
    'in charge," he adds. Trump posted on Truth Social that the US was "in serious '
    'discussions with A NEW, AND MORE REASONABLE, REGIME to end our Military Operations '
    'in Iran". Asked about the post, Rubio says: "If there are new people in charge who '
    'have more of a reasonable vision of the future that would be good news... But we '
    'also have to be prepared for the possibility, maybe even the probability that that '
    'is not the case." Iran has acknowledged messages being passed through Pakistani '
    'intermediaries but denies direct involvement in talks, with a foreign ministry '
    'spokesman calling US proposals \"excessive\".'
)

SENTENCES = [s.strip() for s in re.split(r'(?<=[.!?"])\s+', TEXT) if len(s.strip()) > 20]

print(f"Text length : {len(TEXT)} characters")
print(f"Sentences   : {len(SENTENCES)}\n")
for i, s in enumerate(SENTENCES, 1):
    print(f"  [{i}] {s[:110]}{'…' if len(s)>110 else ''}")
````

## 3. Method 1 — Bayesian Naïve Bayes

Uses a hand-crafted diplomatic/political **lexicon** as word-level likelihood estimates.  
The decision rule is the **log-posterior ratio**: log P(pos|doc) − log P(neg|doc).  
Laplace smoothing (α=1) handles unseen vocabulary.


````python
LEXICON: Dict[str, float] = {
    # Positive tokens
    "reasonable": 1.5, "good": 1.5, "new": 0.8, "vision": 1.0,
    "serious": 0.5, "discussions": 0.3, "acknowledged": 0.5,
    "prepared": 0.5, "future": 0.8,
    # Negative tokens
    "fail": -1.8, "trouble": -1.5, "probability": -0.5,
    "possibility": -0.3, "excessive": -1.5, "denies": -1.2,
    "military": -0.8, "operations": -0.4, "charge": -0.4, "not": -0.6,
}

ALPHA      = 1.0
VOCAB_SIZE = len(LEXICON) + 100

def tokenise(text: str) -> List[str]:
    return re.findall(r'\b[a-z]+\b', text.lower())

def bayesian_sentiment(text: str) -> Dict:
    tokens = tokenise(text)
    log_prior_pos = math.log(0.45)
    log_prior_neg = math.log(0.55)
    ll_pos = ll_neg = 0.0
    for t in tokens:
        sc = LEXICON.get(t, 0.0)
        p_pos = (max(sc, 0) + ALPHA) / (2.0 + ALPHA * VOCAB_SIZE)
        p_neg = (max(-sc, 0) + ALPHA) / (2.0 + ALPHA * VOCAB_SIZE)
        ll_pos += math.log(p_pos)
        ll_neg += math.log(p_neg)
    lp_pos = log_prior_pos + ll_pos
    lp_neg = log_prior_neg + ll_neg
    mx = max(lp_pos, lp_neg)
    ep, en = math.exp(lp_pos - mx), math.exp(lp_neg - mx)
    prob_pos, prob_neg = ep / (ep + en), en / (ep + en)
    label = "POSITIVE" if prob_pos > 0.55 else ("NEGATIVE" if prob_neg > 0.55 else "NEUTRAL")
    return {"label": label, "prob_positive": round(prob_pos, 4),
            "prob_negative": round(prob_neg, 4), "log_odds": round(lp_pos - lp_neg, 4)}

bayes_full = bayesian_sentiment(TEXT)
print("=== Bayesian Sentiment (full text) ===")
for k, v in bayes_full.items():
    print(f"  {k:<16}: {v}")
````

### Bayesian — Visualise Sentence-Level Log-Odds

````python
bayes_sentences = [bayesian_sentiment(s) for s in SENTENCES]
log_odds = [r["log_odds"] for r in bayes_sentences]
colors = [ACCENT[0] if lo >= 0 else ACCENT[3] for lo in log_odds]

fig, ax = plt.subplots(figsize=(11, 4))
bars = ax.bar(range(1, len(SENTENCES)+1), log_odds, color=colors, width=0.6, zorder=3)
ax.axhline(0, color="#ffffff", linewidth=0.8, linestyle="--", alpha=0.4)
ax.set_xlabel("Sentence index")
ax.set_ylabel("Log-odds  (pos vs neg)")
ax.set_title("Bayesian Sentiment — Log-Posterior Ratio per Sentence", fontweight="bold", pad=14)
ax.set_xticks(range(1, len(SENTENCES)+1))
ax.grid(axis="y", zorder=0)
pos_p = mpatches.Patch(color=ACCENT[0], label="Positive")
neg_p = mpatches.Patch(color=ACCENT[3], label="Negative")
ax.legend(handles=[pos_p, neg_p], loc="upper right")
plt.tight_layout()
plt.show()
print(f"\nFull-text Bayesian label → {bayes_full['label']}  "
      f"(P+={bayes_full['prob_positive']:.3f}, P-={bayes_full['prob_negative']:.3f})")
````

## 4. Method 2 — BERT Sentiment (RoBERTa)

Model: `cardiffnlp/twitter-roberta-base-sentiment-latest`  
Outputs: **NEGATIVE / NEUTRAL / POSITIVE** with a confidence score.  
Downloads the model weights on first run (~500 MB).


````python
def bert_sentiment(text: str) -> Dict:
    try:
        from transformers import pipeline
        clf = pipeline("sentiment-analysis",
                       model="cardiffnlp/twitter-roberta-base-sentiment-latest",
                       truncation=True, max_length=512)
        r = clf(text)[0]
        return {"label": r["label"].upper(), "score": round(r["score"], 4),
                "model": "cardiffnlp/twitter-roberta-base-sentiment-latest"}
    except ImportError:
        return {"error": "pip install transformers torch"}
    except Exception as e:
        return {"error": str(e)}

print("Loading BERT sentiment model (may take a moment)…")
bert_full = bert_sentiment(TEXT)
print("\n=== BERT Sentiment (full text) ===")
for k, v in bert_full.items():
    print(f"  {k:<8}: {v}")
````

### BERT — Sentence-Level Scores

````python
print("Running BERT on each sentence…")
bert_sentences = [bert_sentiment(s) for s in SENTENCES]

label_color = {"POSITIVE": ACCENT[2], "NEGATIVE": ACCENT[3], "NEUTRAL": ACCENT[0]}
scores  = [r.get("score", 0) for r in bert_sentences]
labels  = [r.get("label", "N/A") for r in bert_sentences]
colors2 = [label_color.get(lb, "#888") for lb in labels]

fig, ax = plt.subplots(figsize=(11, 4))
ax.bar(range(1, len(SENTENCES)+1), scores, color=colors2, width=0.6, zorder=3)
ax.set_xlabel("Sentence index")
ax.set_ylabel("Confidence score")
ax.set_title("BERT Sentiment — Confidence per Sentence", fontweight="bold", pad=14)
ax.set_xticks(range(1, len(SENTENCES)+1))
ax.set_ylim(0, 1)
ax.grid(axis="y", zorder=0)
handles = [mpatches.Patch(color=c, label=l) for l, c in label_color.items()]
ax.legend(handles=handles, loc="lower right")

for i, (sc, lb) in enumerate(zip(scores, labels)):
    ax.text(i+1, sc+0.02, lb[0], ha="center", fontsize=8, color="white")

plt.tight_layout()
plt.show()
````

## 5. Method 3 — BERT Ekman Emotions

Model: `j-hartmann/emotion-english-distilroberta-base`  
Outputs scores for the **6 Ekman basic emotions** + neutral:  
anger · disgust · fear · joy · sadness · surprise · neutral


````python
def bert_ekman(text: str) -> Dict:
    try:
        from transformers import pipeline
        clf = pipeline("text-classification",
                       model="j-hartmann/emotion-english-distilroberta-base",
                       top_k=None, truncation=True, max_length=512)
        results = sorted(clf(text)[0], key=lambda x: x["score"], reverse=True)
        return {"model": "j-hartmann/emotion-english-distilroberta-base",
                "emotions": [{"label": r["label"], "score": round(r["score"], 4)} for r in results],
                "dominant": results[0]["label"]}
    except ImportError:
        return {"error": "pip install transformers torch"}
    except Exception as e:
        return {"error": str(e)}

print("Loading Ekman emotion model…")
ekman_full = bert_ekman(TEXT)
print("\n=== BERT Ekman Emotions (full text) ===")
if "emotions" in ekman_full:
    print(f"  Dominant : {ekman_full['dominant'].upper()}")
    for em in ekman_full["emotions"]:
        bar = "█" * int(em["score"] * 40)
        print(f"  {em['label']:10s} {em['score']:.4f}  {bar}")
else:
    print(ekman_full)
````

### Ekman — Full-Text Emotion Bar Chart

````python
if "emotions" in ekman_full:
    emo_labels = [e["label"] for e in ekman_full["emotions"]]
    emo_scores = [e["score"] for e in ekman_full["emotions"]]
    emo_colors = ACCENT[:len(emo_labels)]

    fig, ax = plt.subplots(figsize=(9, 4))
    bars = ax.barh(emo_labels, emo_scores, color=emo_colors, height=0.55, zorder=3)
    ax.set_xlim(0, max(emo_scores) * 1.25)
    ax.set_xlabel("Score")
    ax.set_title("BERT Ekman Emotions — Full Text", fontweight="bold", pad=14)
    ax.grid(axis="x", zorder=0)

    for bar, sc in zip(bars, emo_scores):
        ax.text(sc + 0.005, bar.get_y() + bar.get_height()/2,
                f"{sc:.4f}", va="center", fontsize=9)

    plt.tight_layout()
    plt.show()
````

### Ekman — Radar Chart

````python
if "emotions" in ekman_full:
    labels_r = [e["label"] for e in ekman_full["emotions"]]
    scores_r = [e["score"] for e in ekman_full["emotions"]]
    N = len(labels_r)
    angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
    scores_r += scores_r[:1]
    angles   += angles[:1]

    fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
    ax.set_facecolor("#1a1d27")
    ax.plot(angles, scores_r, color=ACCENT[1], linewidth=2)
    ax.fill(angles, scores_r, color=ACCENT[1], alpha=0.25)
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels_r, fontsize=10)
    ax.set_title("Ekman Emotion Radar", fontweight="bold", pad=20, color="#e8ecf4")
    ax.tick_params(colors="#c8ccd8")
    ax.grid(color="#3a3d4d", linewidth=0.8)
    plt.tight_layout()
    plt.show()
````

### Ekman — Sentence-Level Heatmap

````python
print("Running Ekman on each sentence…")
ekman_sentences = [bert_ekman(s) for s in SENTENCES]

# Build matrix
all_emotions = [e["label"] for e in ekman_full.get("emotions", [])]
matrix = np.zeros((len(SENTENCES), len(all_emotions)))
for i, res in enumerate(ekman_sentences):
    if "emotions" in res:
        score_map = {e["label"]: e["score"] for e in res["emotions"]}
        for j, em in enumerate(all_emotions):
            matrix[i, j] = score_map.get(em, 0)

fig, ax = plt.subplots(figsize=(11, 0.55 * len(SENTENCES) + 1.5))
im = ax.imshow(matrix, aspect="auto", cmap="YlOrRd")
ax.set_xticks(range(len(all_emotions)))
ax.set_xticklabels(all_emotions, rotation=30, ha="right", fontsize=9)
ax.set_yticks(range(len(SENTENCES)))
ax.set_yticklabels([f"S{i+1}" for i in range(len(SENTENCES))], fontsize=9)
ax.set_title("Ekman Emotions per Sentence — Heatmap", fontweight="bold", pad=14)
plt.colorbar(im, ax=ax, label="Score")
plt.tight_layout()
plt.show()
````

## 6. Method 4 — GPT Sentiment + Ekman (OpenAI API)

Sends the text to `gpt-4o-mini` with a structured JSON prompt.  
Returns **sentiment label + confidence + all 7 Ekman scores** (including *contempt*).

> ⚠️ Set `OPENAI_API_KEY` in the cell below (or as an environment variable).


````python
# ── Set your key here OR export OPENAI_API_KEY before launching Jupyter ──
import os
# os.environ["OPENAI_API_KEY"] = "sk-..."   # ← uncomment and paste key

GPT_SYSTEM = """You are a precision sentiment and emotion analyst.
For any text, return ONLY a valid JSON object with this schema:
{
  "sentiment": {"label": "POSITIVE"|"NEGATIVE"|"NEUTRAL"|"MIXED",
                "confidence": <float 0-1>, "rationale": "<one sentence>"},
  "ekman_emotions": {"anger":<float>,"disgust":<float>,"fear":<float>,
                     "joy":<float>,"sadness":<float>,"surprise":<float>,"contempt":<float>},
  "dominant_emotion": "<one of the seven above>",
  "tone_notes": "<brief diplomatic/rhetorical tone description>"
}
All emotion scores must sum to ~1.0. Return ONLY the JSON, no markdown."""

def gpt_sentiment_ekman(text: str, model: str = "gpt-4o-mini") -> Dict:
    try:
        import openai
        client = openai.OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))
        resp = client.chat.completions.create(
            model=model, temperature=0.1, max_tokens=600,
            messages=[{"role": "system", "content": GPT_SYSTEM},
                      {"role": "user",   "content": text}])
        raw = resp.choices[0].message.content.strip()
        raw = re.sub(r"^```(?:json)?\s*|\s*```$", "", raw, flags=re.DOTALL).strip()
        return json.loads(raw)
    except ImportError:
        return {"error": "pip install openai"}
    except Exception as e:
        return {"error": str(e)}

print("Calling GPT API…")
gpt_result = gpt_sentiment_ekman(TEXT)
print("\n=== GPT Result ===")
print(json.dumps(gpt_result, indent=2))
````

### GPT — Ekman Emotion Bar Chart

````python
ek = gpt_result.get("ekman_emotions", {})
if ek:
    gpt_labels = list(ek.keys())
    gpt_scores = list(ek.values())
    gpt_colors = ACCENT[:len(gpt_labels)]

    fig, axes = plt.subplots(1, 2, figsize=(13, 4))

    # Bar chart
    axes[0].barh(gpt_labels, gpt_scores, color=gpt_colors, height=0.55, zorder=3)
    axes[0].set_xlim(0, max(gpt_scores) * 1.3)
    axes[0].set_xlabel("Score")
    axes[0].set_title("GPT Ekman Emotions", fontweight="bold", pad=12)
    axes[0].grid(axis="x", zorder=0)
    for j, (sc, lb) in enumerate(zip(gpt_scores, gpt_labels)):
        axes[0].text(sc + 0.005, j, f"{sc:.3f}", va="center", fontsize=9)

    # Radar
    N2 = len(gpt_labels)
    ang2 = np.linspace(0, 2*np.pi, N2, endpoint=False).tolist()
    sc2  = gpt_scores + gpt_scores[:1]
    ang2 += ang2[:1]
    ax2 = fig.add_subplot(122, polar=True)
    ax2.set_facecolor("#1a1d27")
    ax2.plot(ang2, sc2, color=ACCENT[4], linewidth=2)
    ax2.fill(ang2, sc2, color=ACCENT[4], alpha=0.25)
    ax2.set_xticks(ang2[:-1])
    ax2.set_xticklabels(gpt_labels, fontsize=9)
    ax2.set_title("GPT Ekman Radar", fontweight="bold", pad=20, color="#e8ecf4")
    ax2.grid(color="#3a3d4d")
    axes[1].set_visible(False)   # hide the flat axis under polar

    plt.suptitle(
        f"GPT Sentiment: {gpt_result.get('sentiment',{}).get('label','N/A')}  |  "
        f"Dominant: {gpt_result.get('dominant_emotion','N/A').upper()}",
        fontsize=11, fontweight="bold", y=1.02)
    plt.tight_layout()
    plt.show()
    print("\nTone notes:", gpt_result.get("tone_notes",""))
else:
    print("GPT result unavailable — check OPENAI_API_KEY or error above.")
````

````python
print("╔" + "═"*52 + "╗")
print("║  FINAL SENTIMENT COMPARISON SUMMARY               ║")
print("╠" + "═"*52 + "╣")
print(f"║  Bayesian label   : {bayes_full['label']:<32}║")
if "label" in bert_full:
    print(f"║  BERT label       : {bert_full['label']:<32}║")
if "dominant" in ekman_full:
    print(f"║  Ekman dominant   : {ekman_full['dominant'].upper():<32}║")
gpt_s = gpt_result.get("sentiment", {})
if gpt_s:
    print(f"║  GPT label        : {gpt_s.get('label','N/A'):<32}║")
    print(f"║  GPT dominant emo : {gpt_result.get('dominant_emotion','N/A').upper():<32}║")
print("╚" + "═"*52 + "╝")

# Side-by-side bar: Bayesian probabilities
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))
fig.suptitle("Method Comparison — Full Text", fontweight="bold", fontsize=12)

# Bayesian
axes[0].bar(["Positive","Negative"],
            [bayes_full["prob_positive"], bayes_full["prob_negative"]],
            color=[ACCENT[2], ACCENT[3]], width=0.5, zorder=3)
axes[0].set_ylim(0, 1); axes[0].set_title("Bayesian")
axes[0].grid(axis="y", zorder=0)

# BERT confidence
if "score" in bert_full:
    axes[1].bar([bert_full["label"]], [bert_full["score"]],
                color=ACCENT[0], width=0.5, zorder=3)
    axes[1].set_ylim(0, 1); axes[1].set_title("BERT Sentiment")
    axes[1].grid(axis="y", zorder=0)
else:
    axes[1].text(0.5, 0.5, "BERT N/A", ha="center", va="center", transform=axes[1].transAxes)

# GPT confidence
if gpt_s:
    axes[2].bar([gpt_s.get("label","?")], [gpt_s.get("confidence", 0)],
                color=ACCENT[4], width=0.5, zorder=3)
    axes[2].set_ylim(0, 1); axes[2].set_title("GPT Sentiment")
    axes[2].grid(axis="y", zorder=0)
else:
    axes[2].text(0.5, 0.5, "GPT N/A\n(set API key)", ha="center", va="center",
                 transform=axes[2].transAxes, fontsize=9)

plt.tight_layout()
plt.show()
````
