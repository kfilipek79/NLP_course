```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

Here's a combined table with four columns:

| POS Tag | POS Description | DEP Tag | DEP Description |
|---------|-----------------|---------|-----------------|
| NOUN | Noun | nsubj | Nominal subject |
| VERB | Verb | ROOT | Root of the sentence (main verb) |
| ADJ | Adjective | dobj | Direct object |
| ADV | Adverb | det | Determiner |
| DET | Determiner (article) | prep | Prepositional modifier |
| PRON | Pronoun | pobj | Object of preposition |
| ADP | Adposition (preposition) | amod | Adjectival modifier |
| CONJ | Conjunction | advmod | Adverbial modifier |
| PROPN | Proper noun | aux | Auxiliary verb |
| NUM | Numeral | compound | Compound modifier |
| PUNCT | Punctuation | - | - |

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")

for token in doc:
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
            token.shape_, token.is_alpha, token.is_stop)
    
```

```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("This is a sentence.")
displacy.serve(doc, style="dep", port=5001)
```
